## This notebook wil demonstrate a simple Perceptron 

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import  train_test_split
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
from mlxtend.plotting import plot_decision_regions
import seaborn as sns 
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
df = load_iris(as_frame=True)
df = df.frame
df.head()
```

```python
# Keeping only 2 row that is sepal lenght and sepal width 

df = df.drop(["petal length (cm)", "petal width (cm)"], axis = 1)
df.head(3)
```

```python
sns.scatterplot(data= df, x = "sepal length (cm)", y = "sepal width (cm)", hue= "target")
plt.show()
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.drop("target", axis=1), df["target"], test_size=0.2, random_state=42)
```

```python
model = Perceptron()
model.fit(x_train,y_train)
model.coef_, model.intercept_
```

```python
y_pred = model.predict(x_test)
print(accuracy_score(y_pred,y_test))
```

```python
plot_decision_regions(x_train.to_numpy(),y_train.to_numpy(), model)
plt.show()
```
