```python
import numpy as np 
from sklearn.datasets import make_classification
```

```python
x,y = make_classification(n_classes=2, n_samples=100, n_features=2, n_informative=1, n_redundant=0, n_clusters_per_class=1)
x.shape, y.shape
```

```python
def step(z): #activation function 
    return  1 if z>0 else 0 # return the final result based on the input
```

```python
def perceptron(x,y):
    x = np.insert(x, 0,1,axis=1)
    weights = np.ones(x.shape[1]) # creates a 1d array with all the elements as 1 and the same shape as x.shape
    lr = 0.1
    for i in range(1000):
        j = np.random.randint(0,100)
        y_pred = step(np.dot(x[j],weights)) # perceptron summation node  and the prediction function
        weights = weights + lr*(y[j]-y_pred)*x[j] ## formula to update the new weights  
        
    return weights[0], weights[1:]
```

```python
intercept, coef = perceptron(x,y)
intercept,coef
```

## Finding the slope as we have found the intercept and the coef now find the slope and y-intercept 
###  Finding slope from a general linear equation  ax + by + c = 0:


### m  = -(a/b) or -(coef of x / coef of y ) slope
### y-intercept = -(intercept/ceof of y)

```python
m = -(coef[0]/coef[1])
b = -(intercept/coef[1])
```

```python
x_input = np.linspace(-3,3,100) # this is for creating a smooth line 
y_input = m*x_input + b
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,5))
plt.scatter(x[:,0],x[:,1], c = y, cmap= "winter")
plt.plot(x_input, y_input, color = "red")
plt.show()
```
